```python
# ==========================================
# 1. IMPORT LIBRARIES
# ==========================================
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix

import tensorflow as tf
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing import image

# ==========================================
# 2. LOAD AND PREPROCESS CIFAR-10 DATASET
# ==========================================
# Load color image data (3 channels: RGB, size: 32x32 pixels)
(x_train, y_train), (x_test, y_test) = cifar10.load_data()

print("Training data shape:", x_train.shape)  # Expected: (50000, 32, 32, 3)
print("Test data shape:", x_test.shape)      # Expected: (10000, 32, 32, 3)

# Define human-readable labels for the 10 categories
class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer',
               'dog', 'frog', 'horse', 'ship', 'truck']

# Normalize pixel values from [0-255] down to [0.0 - 1.0] range
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Store original labels for evaluation later before one-hot encoding
y_true_train = y_train.flatten()
y_true_test = y_test.flatten()

# One-hot encode the target arrays
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

print("Preprocessed x_train shape:", x_train.shape)
print("Preprocessed y_train shape:", y_train.shape)

# ==========================================
# 3. VISUALIZE TRAIN DATA SAMPLES
# ==========================================
# Plot the first 9 color images from the training set in a 3x3 grid
plt.figure(figsize=(7, 7))
for i in range(9):
    plt.subplot(3, 3, i + 1)
    plt.imshow(x_train[i])  # Color images don't require .reshape() or gray cmaps
    label_index = np.argmax(y_train[i])
    plt.title(f"Label: {class_names[label_index]}")
    plt.axis('off')
plt.tight_layout()
plt.show()

# ==========================================
# 4. DEFINE THE CNN ARCHITECTURE FOR IMAGES
# ==========================================
# Deeper model with dropout to handle complex multi-channel patterns
model = Sequential([
    # Convolutional Block 1
    Conv2D(32, kernel_size=(3, 3), activation='relu', padding='same', input_shape=(32, 32, 3)),
    Conv2D(32, kernel_size=(3, 3), activation='relu'),
    MaxPooling2D(pool_size=(2, 2)),
    Dropout(0.25),

    # Convolutional Block 2
    Conv2D(64, kernel_size=(3, 3), activation='relu', padding='same'),
    Conv2D(64, kernel_size=(3, 3), activation='relu'),
    MaxPooling2D(pool_size=(2, 2)),
    Dropout(0.25),

    # Dense Classification Layers
    Flatten(),
    Dense(512, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')
])

# Display network layout
model.summary()

# ==========================================
# 5. COMPILE AND TRAIN MODEL
# ==========================================
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Train the model
history = model.fit(
    x_train, y_train,
    epochs=10,             # Increased to 10 for better convergence on complex object features
    batch_size=128,
    validation_split=0.1,
    verbose=1
)

# ==========================================
# 6. EVALUATE THE MODEL
# ==========================================
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=0)
print(f"\nTest Accuracy: {test_acc:.4f}")
print(f"Test Loss: {test_loss:.4f}\n")

# ==========================================
# 7. METRICS: REPORT & CONFUSION MATRIX
# ==========================================
# Get prediction classes across the test pool
y_pred = model.predict(x_test)
predicted_classes = np.argmax(y_pred, axis=1)

# Generate Classification Performance Summary Text
print("Classification Report:\n")
print(classification_report(y_true_test, predicted_classes, target_names=class_names))

# Compute Confusion Matrix Layout
conf_matrix = confusion_matrix(y_true_test, predicted_classes)

# Render Confusion Matrix Graphically
plt.figure(figsize=(10, 8))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names, cbar=False)
plt.title("Confusion Matrix - Image Recognition")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.show()

# ==========================================
# 8. BATCH PREDICTION VISUALIZATION (10 TEST IMAGES)
# ==========================================
# Show 10 test predictions side by side horizontally
plt.figure(figsize=(15, 6))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_test[i])
    plt.title(f"Pred: {class_names[predicted_classes[i]]}\nAct: {class_names[y_true_test[i]]}")
    plt.axis('off')
plt.tight_layout()
plt.show()

# ==========================================
# 9. PREDICT A SINGLE OPTIONAL IMAGE FROM TEST SET
# ==========================================
index = 750  # Change index value to check any test image sample

# Add required 4D batch tensor axis via np.expand_dims -> shape becomes (1, 32, 32, 3)
single_image = np.expand_dims(x_test[index], axis=0)

single_prediction = model.predict(single_image)
pred_idx = np.argmax(single_prediction[0])
act_idx = y_true_test[index]

plt.figure(figsize=(3, 3))
plt.imshow(x_test[index])
plt.title(f"Single Test Pred: {class_names[pred_idx]}\nActual: {class_names[act_idx]}")
plt.axis('off')
plt.show()

# ==========================================
# 10. PREDICT A CUSTOM EXTERNAL USER-UPLOADED IMAGE
# ==========================================
# Make sure to replace 'your_custom_image.jpg' with a real file in your local directory!
img_path = 'your_custom_image.jpg'

try:
    # Safely load the external image file and force-resize to 32x32 pixels
    img = image.load_img(img_path, target_size=(32, 32))

    # Transform raw image pixels into matrix format
    img_array = image.img_to_array(img)

    # Scale intensities to [0, 1] matching training configuration
    img_array = img_array.astype('float32') / 255.0

    # Expand to 4D batch array -> shape (1, 32, 32, 3)
    img_tensor = np.expand_dims(img_array, axis=0)

    # Predict the target output
    custom_pred = model.predict(img_tensor)
    custom_idx = np.argmax(custom_pred[0])
    confidence = custom_pred[0][custom_idx] * 100

    # Plot custom image prediction result
    plt.figure(figsize=(3, 3))
    plt.imshow(img)
    plt.title(f"External File Pred: {class_names[custom_idx]}\nConfidence: {confidence:.2f}%")
    plt.axis('off')
    plt.show()

except FileNotFoundError:
    print(f"\n[Notice]: Custom file '{img_path}' was not found. Section 10 was skipped.")
    print("To test section 10, save any picture in your workspace folder and rename the 'img_path' variable string above to match it.")
```
